```python
import numpy as np
```

We compute the matrix multiplication

$$
\begin{bmatrix}
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
\begin{bmatrix}
1 \\
2 \\
3
\end{bmatrix}.
$$

The dimensions are $(2 \times 3)(3 \times 1)$, so the result is a $(2 \times 1)$ vector.

First row:
$$
0\cdot 1 + 1\cdot 2 + 0\cdot 3 = 2.
$$

Second row:
$$
0\cdot 1 + 0\cdot 2 + 1\cdot 3 = 3.
$$

Therefore,

$$
\begin{bmatrix}
2 \\
3
\end{bmatrix}.
$$

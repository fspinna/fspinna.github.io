Answer: because **constant values can be pooled**.

It is an optimization strategy by the interpreter.

Some *immutable objects*, such as strings and numbers, **can be reused to save memory.**

For example, when you create a string, Python may decide to use an existing version of that string, rather than creating a new one. This is called "**string interning.**"

So, if you have two variables that contain the same string, Python may refer to the same memory location for both, making memory allocation more efficient.

# Functions

A function defines a block of code that can be reused<br>
A function has a number of input arguments and a return value.

```python
def oldest(name_age_dict):
    """
    Returns the name of the oldest person in the dictionary
    The dictionary must map names to ages

    :param name_age_dict: map names to ages
    :return: returns the name of the oldest person in the dictionary
    """
    # the string above is a docstring, a human readable description of the function
    max_age = -1
    oldest_name = None
    for name in name_age_dict:
        if name_age_dict[name] > max_age:
            max_age = name_age_dict[name]
            oldest_name = name
    return oldest_name
```

Input arguments takes values in the same way of assignment<br>
You can define default value in case an argument is not specifies

```python
def oldest(name_age_dict, noname='noname'):
    """
    Returns the name of the oldest person in the dictionary
    The dictionary must map names to ages

    :param name_age_dict: map names to ages
    :param noname: default value if no names are passed
    :return: returns the name of the oldest person in the dictionary
    """
    # the string above is a docstring, a human readable description of the function
    max_age = -1
    oldest_name= None
    for name in name_age_dict:
        if name_age_dict[name]>max_age:
            max_age = name_age_dict[name]
            oldest_name = name
    if oldest_name is None:
        return noname
    else:
        return oldest_name
```

A function that takes two arguments which have default values.

The function uses the `randint` function to extract a random number witin a range of integers.

```python
import random
def roll_dices(number_of_dices = 1, faces = 6):
    sum = 0
    for i in range(number_of_dices):
        sum += random.randint(1, faces)
    return sum
```

Invoking the function returns a possibly different value each time.

```python
roll_dices()
```

```python
roll_dices(2,12)
```

If you want to use the default value for an argument and set the second one,
pass the second argument with its name

```python
roll_dices(faces=9) # rolls one dice with nine faces
```

# generator

Generators are a way to define an iterator that computes its elements *on demand*.<br>
You can turn a function into a generator by using operator `yield` to compute successive values.

`yield`, like `return`, returns a value to the caller.

While functions return just once, a generator will return a value multiple times, whenever it performs a ` yield`, and continues from there when `next()` is called on it.

```python
def infinite():
    i = 0
    while True:
        yield i # this is where the generator produces its values
        i += 1
```

```python
N = infinite() # invoking the function creates the generator
N # yes, it is an object
```

```python
next(N), next(N), next(N)
```

The functions defining generators can have arguments, just like any other function

```python
def infinite(start = 0, step = 1):
    i = start
    while True:
        yield i
        i += step
N = infinite(10,5)
next(N), next(N), next(N)
```

# List comprehension

A concise way to build lists, out of epxressions that compute the value of each item.

```python
a = list()
for x in range(6):
    a.append(x**2)
a
```

Instead of using an explicit loop, the same list can be build by *list comprehension*, in a more compact and math-like notation:

```python
a = [x**2 for x in range(6)]
a
```

```python
a = list()
for x in range(6):
    if x%2 == 0: # filtering condition
        a.append(x**2)
```

A filtering condition can be added to select specific items

```python
a = [x**2 for x in range(6) if x%2 == 0]
a
```

# Classes

A class wraps in a single entity a state, defined by a set of variables (fields),
and functions to operate on that state (methods)

```python
class Person:
    def __init__(self, name, age): # constructor
        self.name = name          # instance variables
        self.age = age            # self is like 'this' in Java, C#

    def young(self):              # method that return a value
        return self.age < 40

    def birthday(self):           # method that changes the state
        self.age += 1             # of the object
```

The special method `__init__` is invoked when an object is constructed

```python
me = Person('Laura', 40)
```

```python
me.young()
```

```python
me.birthday()
```

```python
me.age
```

Why there is a `self` argument in methods?<br>
`self` is a reference to the instance of the object, similar to `this` in C++, Java, C#<br>
Differently from those languages, in python `self` must be explicitly mentioned.

When you write `me.young()`, `me` is passed automatically as `self` to the method of the class.<br>
In fact you can invoke methods of a class using this equivalent notation:

```python
Person.young(me)
```

You could... but almost no one actually uses this syntax.<br>
It is just good to know.

Classes can have class-wide variables (somewhat like `static` in Java)

```python
class Person(object):
    population = 0                # class variable

    def __init__(self, name, age):
        self.name = name
        self.age = age
        Person.population += 1    # incremented when created

    def young(self):
        return self.age < 40

    def birthday(self):
        self.age += 1

    def __del__(self):
        Person.population -= 1     # decremented when created
```

We use this function to play with class inheritance

```python
def is_young(some_person):
    if some_person.young():
        print(some_person.name, 'is a young person')
    else:
        print(some_person.name, 'was a young person')
```

```python
me = Person('Laura', 39)
is_young(me) # <40 -> young
```

```python
me.birthday() # 40!
print(me.age)
is_young(me) # ==40 -> not young
```

```python
Person.population
```

```python
del me
```

When defining a class you can specify one or more base classes from which
the new class inherits methods and class variables.

Any method with the same name of a base class method hides that base class method.

```python
class Researcher(Person):
    def __init__(self, name, age, discipline):
        super().__init__(name, age) # the super method gives access to base class methods
        self.discipline = discipline

    def __repr__(self):
        print(self.name)
        return self.name

    def young(self): # a researcher is always young because is always learns new things
        return True
```

```python
me = Researcher('Laura', 34, 'SNA')
me.birthday()
me.young() # :)
```

## Dealing with files

The function `open` opens a file both for writing and/or reading
in either text format or in binary format.<br>
The default mode is reading a text file.

Let's open a text file and write two lines in it.

```python
file = open('filename.txt', mode='w', encoding='utf-8') # /content/filename.txt
file.write('ciao\n')   # newlines are not added automatically
file.write('hello\n')
file.close()           # remember to close files
```

Now open the same file to read from it

```python
file = open('filename.txt', encoding='utf-8')
for idx, line in enumerate(file):
  print(idx, line.strip())
```

```python
file = open('filename.txt', encoding='utf-8')
print(next(file))          # a text file object acts as a iterator over lines
```

```python
print(next(file))
```

```python
file.close() # rember to close
```

Rather than remembering to close a file, it is better to close the file automatically,
when it is no longer needed.<br>
This is done using the `with` statement.<br>
The file will close even when an uncatched exception happens.

```python
with open('filename.txt', mode='w', encoding='utf-8') as file:
    file.write('ciao\n')
    file.write('hello\n')
```

```python
with open('filename.txt', encoding='utf-8') as file:
    for line in file:
        print(line.strip())
```

You can read a whole text file in a single string using the read() method

```python
with open('filename.txt', encoding='utf-8') as file:
    text = file.read()
text
```

## CSV files

Files in comma separated value (CSV) format are useful to store tabular data.

```python
import csv
data = [['Laura', 35], ['Gianmarco', 31], ['Paola', 63]]
with open('data.csv', mode='w', encoding='utf-8', newline='') as outfile:
    writer = csv.writer(outfile) # once you open the file you wrap it with a csv.writer or a csv.reader
    writer.writerows(data) # then you can write rows in a single shot...

    # or a row at a time
    for row in data:
      writer.writerow(row)
```

```python
import csv
data = list()
with open('data.csv', encoding='utf-8', newline='') as infile:
    reader = csv.reader(infile)
    for row in reader:
        data.append([row[0],int(row[1])])
data # data is duplicate beacuse we wrote it twice in the previous cell
```

The Pandas module allows to perform complex manipulation of tabular data, in the form of dataframes.

It support loading and writing data in csv files, and many other formats.

```python
import pandas as pd

df = pd.read_csv('data.csv', names=['name','age']) # 'names' define the columns names, when the input file does not define them.
df
```

# Regular Expressions in Python

This notebook introduces the basics of regular expressions (regex) in Python.

````python
import re
````

`findall()` scans text left to right and, after each match, resumes at the match’s end, advancing one position after an empty match. Regex quantifiers are greedy by default, so it matches the longest pattern.

## Concatenation
Patterns can be concatenated simply by placing them next to each other. For example, `abc` matches the substring `'abc'` exactly.

````python
text = "abcdef"
print(re.findall(r"abc", text))  # matches 'abc'
print(re.findall(r"abcghhjjjknknikhhnk", text))  # matches 'abc'
````

## Anchors
`^` anchors the pattern to the beginning of the string.
`$` anchors the pattern to the end of the string.

````python
print(re.findall(r"^abc", "abcdef"))   # matches at start
print(re.findall(r"^abc", "defabc"))   # matches at start

print(re.findall(r"def$", "abcdef"))  # matches at end
print(re.findall(r"def$", "defabc"))  # matches at end
````

## Quantifiers
- `*` zero or more
- `+` one or more
- `?` zero or one
- `{m,n}` between m and n times

````python
print(re.findall(r"a*", "caaaat"))   # zero or more 'a'
print(re.findall(r"a+", "caaaat"))   # one or more 'a'
print(re.findall(r"a?", "caaaat"))   # zero or one 'a'
print(re.findall(r"a{2,3}", "caaaat")) # 2 to 3 'a'
````

## Alternation
The pipe `|` operator matches one of several patterns.

````python
print(re.findall(r"cat|dog", "I have a cat and a dog"))
print(re.findall(r"cat|dog", "I have a duck and a dog"))
````

## Character Classes and Metacharacters
Square brackets `[]` define a set of characters to match.
- `[abc]` matches `a`, `b`, or `c`.
- `[0-9]` matches any digit.
- Metacharacters like `.`, `\d`, `\w`, `\s` provide shortcuts.

````python
print(re.findall(r"[aeiou]", "hello world"))   # vowels
print(re.findall(r"[0-9]", "a1b2c3.,+"))         # digits
print(re.findall(r".", "abc"))                # any character
print(re.findall(r".", "12+-*()ABCa "))                # any character
````

## Grouping
Parentheses `()` group patterns together and create capture groups.

````python
text = "Hello World"
print(re.sub(r"(\w+) (\w+)", r"\2, \1", text))
# → 'World, Hello'
````

## Negation
A caret `^` inside `[]` negates the character class.

````python
print(re.findall(r"[^aeiou]", "hello"))  # non-vowels
````

## Escape Sequences, Ranges, Advanced Operators
- `\d` matches digits `[0-9]`
- `\D` matches non-digits
- `\w` matches word characters `[a-zA-Z0-9_]`
- `\W` matches non-word characters
- `\s` matches whitespace
- `\S` matches non-whitespace

````python
sample = "User123 logged in!"
print(re.findall(r"\d+", sample))   # digits
print(re.findall(r"\D+", sample))   # non-digits
print(re.findall(r"\w+", sample))   # words
print(re.findall(r"\W+", sample))   # spaces and symbols
print(re.findall(r"\s", sample))    # whitespace
print(re.findall(r"\S+", sample))   # non-whitespace groups
````

## Substitutions and Capture Groups
Use `re.sub(pattern, replacement, string)` to replace text.
Captured groups can be referenced in the replacement using `\1`, `\2`, etc.

````python
text = "item 42"
print(re.sub(r"([a-z]+) (\d+)", r"<\2> <\1>", text))
# → '<item> <42>'
````

## Lookahead
Lookahead assertions check if a pattern exists ahead without consuming characters.
- Positive lookahead: `(?=...)`
- Negative lookahead: `(?!...)`

````python
text = "foo123 bar"

# positive lookahead: match 'foo' only if followed by digits
print(re.findall(r"foo(?=\d+)", text))

# negative lookahead: match 'bar' only if not followed by digits
print(re.findall(r"bar(?!\d)", text))
print(re.findall(r"foo(?!\d)", text))
````
